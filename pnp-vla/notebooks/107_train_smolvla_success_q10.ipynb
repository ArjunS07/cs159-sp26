# 107 — SmolVLA success-probability Q10 training

This is a new objective-aligned experiment; it does not resume or overwrite notebooks 94–95. It trains two matched critics from scratch on the same immutable 480-tree v3 snapshot. Root MC learns observed success after one ten-action intervention followed by the frozen P&P policy. Tree TD uses binary success reward, gamma=1, and the recorded next action under that continuation policy. Both see normalized remaining episode time. No demonstrator TD transitions enter either objective.

Run root MC and tree TD sequentially here, or select one arm per Colab runtime. The output directories are separate and resumable. Compare them in notebook 106. These experiments estimate selection quality on held-out roots; they are not a closed-loop success claim.
