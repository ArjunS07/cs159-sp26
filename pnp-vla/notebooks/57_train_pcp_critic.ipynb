Thin GPU launcher. It builds/resumes a compact local cache one rollout at a time, then trains with a bounded two-rollout RAM cache. The checkpoint is registered offline-only; this does not modify rollout behavior.

```python
from pnp.pcp_critic import PCPCriticTrainConfig, run_pcp_critic_train

SNAPSHOT_ID = 'PASTE_PCPCDS_SNAPSHOT_ID'
OBJECTIVE = 'calql'  # repeat with 'iql' for the matched ablation
CACHE_ROOT = '/content/pcp_critic_cache'  # local disk; safe to reuse after a restart
assert SNAPSHOT_ID.startswith('pcpcds-')
CONFIG = PCPCriticTrainConfig(objective=OBJECTIVE, updates=10_000, batch_size=64)
CRITIC_ID = run_pcp_critic_train(
    snapshot_id=SNAPSHOT_ID, objective=OBJECTIVE, train_config=CONFIG, cache_root=CACHE_ROOT)
print(CRITIC_ID)
```
