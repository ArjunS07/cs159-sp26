# 88 — SmolVLA depth-1 trees, worker 1/2

Collects 400 fixed roots from notebook 87's indices 10–29 cohort in the fresh `v2-egl` namespace. Each tree reuses the exact stored source continuation and generates eight counterfactuals in one fixed batch: four fresh-initial-noise plus four fixed-initial-noise/new-P&P-perturbation candidates. Roots use the frozen 65% pair-weighted-U10 / 35% uniform manifest. Every completed tree is resume-safe; no v1 tree can be resumed.

```python
# Install the NVIDIA EGL userspace library before importing MuJoCo.
import re, subprocess
driver = subprocess.check_output(
    ['nvidia-smi', '--query-gpu=driver_version', '--format=csv,noheader'],
    text=True).splitlines()[0].strip()
package = f"libnvidia-gl-{re.match(r'\d+', driver).group()}"
print({'driver': driver, 'installing': package})
subprocess.run(['apt-get', 'update', '-qq'], check=True)
policy = subprocess.check_output(['apt-cache', 'policy', package], text=True)
if 'Candidate: (none)' in policy:
    raise RuntimeError(f'{package} is unavailable in this runtime')
subprocess.run(
    ['apt-get', 'install', '-y', '--no-install-recommends', package], check=True)
subprocess.run('ldconfig -p | grep libEGL_nvidia', shell=True, check=True)
```
