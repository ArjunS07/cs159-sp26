Use one copy per shard (`SHARD_INDEX` = 0, 1, or 2). The three shards cover 18 existing source roots. Each root keeps the exact stored stock chunk, adds one fresh-noise P&P chunk, and adds twelve smooth translation interventions: positive and negative offsets on x/y/z at normalized scales 0.02 and 0.06. The frozen P&P continuation and seed are shared. The worker writes full branch artifacts to Supabase under a **new experiment** and skips complete groups when restarted.

This is a small scale-finding pilot, not a held-out evaluation set. Run one root first (`TREE_LIMIT=1`); then set `TREE_LIMIT=None` for six roots in the shard. You start collection by setting `RUN_COLLECTION=True`.
