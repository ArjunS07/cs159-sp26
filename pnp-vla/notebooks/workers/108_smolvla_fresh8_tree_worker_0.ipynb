Collects one fixed shard of 266 or 267 roots from the 800 completed source episodes (LIBERO episode indices 10–29). The frozen manifest chose 65% of roots by high U10 uncertainty and 35% uniformly. Each tree has the exact stored source continuation plus eight fresh-initial-noise P&P chunks. Complete v3 roots reuse fresh seeds 1–4 and collect only seeds 5–8; other roots collect all eight. New branches execute ten actions, then follow the same frozen P&P policy. This is a new v4 namespace; v3 hybrid trees stay untouched. The first run collects one tree; set `TREE_LIMIT = None` for the full shard. Restart with the same shard index to resume completed trees.
