# 35 - Suffix sensitivity and tapered refinement, worker 0/4

This is shard 0 of 4 over 220 held-out LIBERO-PRO identities (11 suites x 10 tasks x init indices 10 and 11). Every identity runs three paired arms: exact stock-action diagnostic baseline, ordinary full P&P refinement, and tapered P&P refinement.

The diagnostic independently resamples only action positions 10–49 and records how the predicted first 10 actions change. It also stores the full 50-position uncertainty profile, so first-10, first-20, and full-chunk uncertainty can be analyzed later. Tapered refinement applies full feedback on positions 0–9, linearly decays across 10–19, and applies no feedback on 20–49 inside every P&P iteration.

Progress prints every 30 rollouts: ten complete three-arm groups. Worker 0 may set `EPISODE_LIMIT = 1` once for a three-rollout smoke test, then restore `None`; completed rows resume safely.
